```python
blank_report = {}

for column in text_like_columns:

    blank_mask = (
        df_clean[column]
        .astype("string")
        .str.fullmatch(r"\s*", na=False)
        & df_clean[column].notna()
    )

    blank_count = blank_mask.sum()

    blank_report[column] = blank_count

    df_clean.loc[blank_mask, column] = pd.NA

print("Empty strings converted into missing values:")

display(
    pd.Series(blank_report, name="Converted_Blanks")
)
```

```python
print("Remaining blank strings:")

for column in text_like_columns:

    remaining = (
        df_clean[column]
        .astype("string")
        .str.fullmatch(r"\s*", na=False)
        & df_clean[column].notna()
    )

    if remaining.sum() > 0:
        print(column, remaining.sum())

print("\nUpdated Missing Values:")

display(df_clean.isna().sum())
```

## Identify All Invalid Ages

To identify negative ages, ages above 120, and decimal ages.

### Decision
Invalid age values will be converted into missing values instead of guessing their correct values.

```python
age_columns = ["suspect_age", "victim_age"]

for col in age_columns:
    negative = (df_clean[col] < 0).sum()
    above_120 = (df_clean[col] > 120).sum()
    decimal = (
        df_clean[col].notna()
        & (df_clean[col] % 1 != 0)
    ).sum()

    print(f"\n{col}")
    print("Negative ages:", negative)
    print("Above 120:", above_120)
    print("Decimal ages:", decimal)
```

```python
#Handle Invalid Age Value

for col in age_columns:
    invalid_age = (
        (df_clean[col] < 0)
        | (df_clean[col] > 120)
    )

    df_clean.loc[invalid_age, col] = float("nan")
```

```python
#decimal age correction

for col in age_columns:
    df_clean[col] = df_clean[col].round()
    
```

```python
# missing age correction

for col in age_columns:
    median_age = df_clean[col].median()
    df_clean[col] = df_clean[col].fillna(median_age)
```
